```python
import random

# Part A – Machine Initialization
machines = ['A', 'B', 'C']
epsilon = 0.2
N_play = 10

# Part B – Initialize N, R and Q tables
N = {m: 0 for m in machines}      # Participation count
R_total = {m: 0 for m in machines} # Total rewards
Q = {m: 0.0 for m in machines}     # Estimated future reward

print("Initial Tables")
print("N =", N)
print("R =", R_total)
print("Q =", Q)
print("-" * 50)

total_reward = 0

for play in range(1, N_play + 1):

    r = random.random()

    # Part C – ε-Greedy Decision
    if r < epsilon:
        selected_machine = random.choice(machines)
        decision = "Exploration"
    else:
        max_q = max(Q.values())
        best_machines = [m for m in machines if Q[m] == max_q]
        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    print(f"\nPlay {play}")
    print(f"Random number = {r:.3f}")
    print(f"{decision}: Selected Machine {selected_machine}")

    # Part D – Reward Collection
    reward = float(input(f"Enter reward obtained from machine {selected_machine}: "))

    # Update counts and rewards
    N[selected_machine] += 1
    R_total[selected_machine] += reward
    total_reward += reward

    # Part E – Q-Value Update
    Q[selected_machine] = Q[selected_machine] + \
                          (reward - Q[selected_machine]) / N[selected_machine]

    print("Updated Q-value =", round(Q[selected_machine], 2))

# Part F – Display Results
print("\n" + "=" * 50)
print("FINAL RESULTS")
print("=" * 50)

print("\nParticipation Count (N)")
for m in machines:
    print(f"{m}: {N[m]}")

print("\nTotal Rewards (R)")
for m in machines:
    print(f"{m}: {R_total[m]}")

print("\nEstimated Q-Values")
for m in machines:
    print(f"{m}: {Q[m]:.2f}")

print("\nTotal Reward Received =", total_reward)
```
