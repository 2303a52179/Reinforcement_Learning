```python
import random

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of machines and plays
M = 3
N_play = 10
epsilon = 0.2

# Part B – Initialize N, R and Q tables
N = {'A': 0, 'B': 0, 'C': 0}       # Participation count
R = {'A': [], 'B': [], 'C': []}     # Rewards received
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0} # Estimated reward

total_reward = 0

print("Modified Epsilon-Greedy Multi-Armed Bandit")
print("Machines:", machines)
print("Number of plays:", N_play)
print("Epsilon:", epsilon)
print("-" * 50)

# Part C – Modified ε-Greedy Decision
for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Highest estimated reward machine
    highest_machine = max(Q, key=Q.get)

    if r < epsilon:
        # Exploration:
        # Select randomly except the highest Q-value machine
        exploration_machines = [
            machine for machine in machines
            if machine != highest_machine
        ]

        selected_machine = random.choice(exploration_machines)
        decision = "Exploration"

    else:
        # Exploitation:
        # Select machine with highest Q-value
        selected_machine = highest_machine
        decision = "Exploitation"

    # Part D – Reward Collection
    print(f"\nPlay {play}")
    print(f"Random number = {r:.4f}")
    print(f"Decision = {decision}")
    print(f"Play with machine {selected_machine}")

    reward = float(input("Enter reward: "))

    # Part E – Q-Value Update
    N[selected_machine] += 1
    R[selected_machine].append(reward)

    # Incremental sample-average formula:
    # Q_new = Q_old + (Reward - Q_old) / N
    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    total_reward += reward

    print(f"Updated Q({selected_machine}) = "
          f"{Q[selected_machine]:.2f}")


# Part F – Display Results
print("\n" + "=" * 60)
print("FINAL RESULTS")
print("=" * 60)

print("\nMachine\tN (Count)\tRewards\t\tQ-Value")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t\t"
        f"{R[machine]}\t"
        f"{Q[machine]:.2f}"
    )

print("\nTotal Reward =", total_reward)
print("Average Reward =", total_reward / N_play)
```
